# From SQL to Spreadsheets

## Introduction: Why Excel Still Rules the Accounting World

Throughout Unit 3, you developed a rigorous foundation in Structured Query Language (SQL). You learned how to retrieve specific records (`SELECT`), link relational tables (`JOIN`), filter row subsets (`WHERE`), group transactions into summary metrics (`GROUP BY`), and rank results (`ORDER BY`). 

Now, we turn our attention to Microsoft Excel. 

At first glance, transitioning from SQL code back to an Excel spreadsheet might feel like taking a step backward. In fact, many students initially wonder: *"If SQL is more structured, auditable, and powerful for large datasets, why do we need Excel?"*

The answer lies in the reality of modern corporate accounting. 

```{note}
**The Industry Reality: Why Public Accounting Firms Demand Excel Power Users**
During a recent visit to a major accounting firm, accounting faculty met with firm partners and senior managers. Following those discussions, the department chair sent an email to the faculty emphasizing a clear message from industry:

> *"During the course of these discussions, it came up that Excel is still VERY important. One of our recent graduates shared that she did not have strong Excel skills prior to her internship. Anything you can do to get students using Excel is greatly appreciated."*

Whether you enter public accounting audit, tax consulting, corporate financial planning and analysis (FP&A), or forensic accounting, Excel remains the primary canvas where business decisions are modeled, financial statements are reviewed, and ad-hoc client requests are solved.
```

However, calling this unit an "Excel unit" is a gross oversimplification. Our goal is **not** to teach basic spreadsheet mechanics—such as entering numbers, formatting fonts, or writing simple addition formulas. You already possess basic Excel skills. Instead, Unit 4 serves two deeper purposes:

1. **Transferring Your SQL Analytical Framework into Spreadsheets:** Demonstrating that everything you learned in SQL provides the underlying logic for how to structure, model, and audit analyses in Excel.
2. **Elevating You to an Excel Power User:** Teaching modern, advanced Excel techniques through real-world business case studies, contrasting crude "old-school" habits with clean, efficient, and auditable spreadsheet architecture.

---

## The Core Philosophy: Analytics is Independent of Software

The central thesis of this textbook is that **data analytics is a process of reasoning, not a collection of software buttons.**

When you mastered SQL, you didn't just learn syntax; you learned **a way of thinking and organizing data analyses**. That same mental framework maps directly onto Microsoft Excel:

| Fundamental Analytical Operation | SQL Clause / Concept | Microsoft Excel Equivalent |
| :--- | :--- | :--- |
| **Data Transformation** | `SELECT UnitPrice * 1.10 AS NewPrice` | Excel Formulas & Column Calculations (`=[@UnitPrice]*1.10`) |
| **Specifying Data Source** | `FROM TableName` | Selecting a Cell Range or Excel Table (`Table1`) |
| **Merging Datasets** | `INNER JOIN / LEFT JOIN ... ON` | Lookup Functions (`XLOOKUP`, `INDEX/MATCH`) & Data Model |
| **Filtering Rows** | `WHERE FilterCondition` | AutoFilter, `FILTER()` function, & Slicers |
| **Grouping & Aggregation** | `GROUP BY Category` | Pivot Tables & Conditional Sums (`SUMIFS`) |
| **Filtering Aggregates** | `HAVING SUM(Sales) > 10000` | Pivot Table Value Filters |
| **Sorting Output** | `ORDER BY SortColumn` | Worksheet Sorting & `SORT()` Dynamic Array Function |

### The "Siren Song" of Excel: Avoiding Haphazard Clicking

Why did we start the semester with SQL instead of jumping straight into Excel?

Excel's graphical interface is a double-edged sword. Because you can see cells on screen, you don't have to write formal computer code to get a quick answer. This convenience creates a dangerous temptation—what data analysts call **the Siren Song of Excel**. 

Inexperienced accountants often open a spreadsheet and begin clicking around haphazardly, copying and pasting numbers, dragging formulas across columns, and deleting cells through trial and error without understanding what they are doing. This leads to broken cell references, hardcoded numbers buried inside formulas, and corrupted audit trails.

Writing SQL **forces** you to write out your step-by-step logic in plain text before you run your code. 

In Unit 4, we will bring that same SQL discipline into Excel. Before clicking a button or typing a formula, you should pause, formulate your logical plan on paper, understand the underlying data structure, and execute your analysis intentionally.

---

## Our Pedagogical Approach: Business Case Studies & Modern Solutions

To extend your Excel proficiency beyond basic mechanics, this unit adopts a problem-driven learning strategy:

1. **Starting with a Real Business Problem:** Every major topic begins with an authentic accounting scenario—such as auditing fixed asset depreciation schedules (PP&E), analyzing credit card expense compliance, or building multi-period sales forecasts.
2. **Exposing Crude vs. Modern Techniques:** We will examine the fragile, "crude" methods that untrained accountants typically use (such as scrolling through thousands of rows, copying and pasting filtered subsets, or chaining nested `VLOOKUP`s), and contrast them directly with modern power-user solutions.

```{warning}
**Old-School Habits vs. Modern Excel Architecture**
Traditional spreadsheet courses taught students to treat Excel as a static grid of manual cells. Modern Excel power users treat spreadsheets as dynamic data pipelines:

* **Old Way:** Manual cell ranges (`A2:D500`), manual copy-pasting, legacy `VLOOKUP`, static formulas, and unorganized sheets.
* **Modern Way:** Structured **Excel Tables**, Dynamic Array formulas (`FILTER`, `SORT`, `UNIQUE`), modern `XLOOKUP`, Pivot Tables, and automated Power Query data connections.
```
