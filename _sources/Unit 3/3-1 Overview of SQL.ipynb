# Overview of SQL

Structured Query Language (SQL) is a programming language designed to interact with relational databases. SQL can retrieve, insert, modify, and delete data from databases. SQL can also change the structure of databases. The following paragraphs elaborate on these capabilities.

## Querying (Retrieving Data)

The word *query* means to ask. When you query a database, you ask it for information. Specifically, you ask it for rows of data from the database. Sometimes, you will retrieve rows directly from a table. In other cases, you will construct rows using information from multiple tables.

Most of your daily workflow will involve *querying*, or fetching, data to find the answers you need. SQL is the perfect tool for this because it simplifies the retrieval process and provides a practical, straightforward approach to accessing information.

### The Pillars of Data Retrieval

To create a useful report from a database, you will typically perform one or more of the following actions:

1. **Specifying the data source**: This is the starting point where you identify which table contains the raw information you need.
2. **Merging**: In accounting, data is often spread out. This step allows you to link tables together. For example, connecting a list of transactions to a list of vendor names so your report shows names instead of just ID numbers.
3. **Filtering**: This allows you to restrict the data to rows of interest. You can instruct SQL to only show records that meet specific criteria, such as *"only invoices that are more than 30 days overdue."*
4. **Selecting**: Databases often have dozens of columns you do not need. This step allows you to pick only the specific fields, like "Date," "Amount," and "Account Name", that belong in your report.
5. **Transforming**: In this step, you can change the data that you retrieve. For example, you can perform math, such as calculating a 5% discount or computing the tax on an item.
6. **Aggregating**: This is where you summarize. Instead of looking at 5,000 individual sales, you can use this pillar to see the total sales per department or the average transaction value.
7. **Sorting**: Finally, you want your data to be readable. This step organizes your results in a logical sequence, such as listing your largest expenses at the top or arranging entries chronologically.

The following SQL code will query the *Northwind Traders* database. It will list all customers located in Portland, Oregon, USA. For each customer, it will count the number of orders that the customer has placed with *Northwind Traders*.

| SQL Code | Pillar |
| -------- | ------ |
| SELECT<br>&nbsp;&nbsp;&nbsp;C.CustomerID,<br>&nbsp;&nbsp;&nbsp;C.CompanyName,<br>&nbsp;&nbsp;&nbsp;COUNT(*) AS [Number of Orders] | Selecting |
| FROM Orders AS O | Specifying the data source |
| INNER JOIN Customers AS C<br>ON O.CustomerID = C.CustomerID | Merging |
| WHERE C.City = 'Portland' | Filtering |
| GROUP BY C.CustomerID, C.CompanyName | Aggregating |
| ORDER BY [Number of Orders] DESC | Sorting |

Do not expect to understand the details of this code. Instead, read the code with a "birds eye" view. Notice how the SQL code is fairly readable and understandable.
