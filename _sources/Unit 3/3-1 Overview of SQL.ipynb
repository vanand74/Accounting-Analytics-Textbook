# Overview of SQL

## What is SQL?

Structured Query Language (SQL) -- pronounced as either "S-Q-L" or "sequel" -- is a programming language designed to interact with relational databases. SQL differs from many mainstream programming languages. In SQL, you describe what data you want to retrieve and how you want to transform it. The database management system then handles the low-level details under the hood. SQL is also unique because it reads like English.

SQL has extensive capabilities. It can:
* *Retrieve Data:* Extract specific rows, columns, and aggregated summaries from tables.
* *Modify Data:* Insert new records, update existing records, or delete records.
* *Define and Modify Database Structure:* Create new tables, alter column definitions, set data constraints, or establish primary and foreign keys.
* *Control Access:* Grant or revoke user permissions to secure sensitive financial records.

## Our Focus: Data Retrieval

While SQL can perform all of the functions listed above, your daily responsibility as an accountant, financial analyst, or auditor will focus almost exclusively on **data retrieval.**

In corporate accounting, data integrity and system security are paramount. Enterprise resource planning (ERP) systems -- such as SAP, Oracle, or NetSuite -- manage live operational data. Allowing staff accountants or external auditors to modify or delete data in production databases poses immense operational risk. A single incorrect command could alter historical general ledger balances or corrupt transaction logs.

For this reason, accounting professionals and auditors are typically granted **read-only access** to databases. By focusing exclusively on retrieving data using the SQL **SELECT** statement, we emulate the exact procedures used by external auditors and forensic accounting teams to independently extract and verify financial data in the field.

### Defining a Query

In computer science, the word **query** literally means to ask. When you write a SQL statement to retrieve data, you are querying the database -- asking it a specific business question, such as:

* "*Which corporate credit card transactions exceeded $10,000 this month?*"
* "*What were our total sales by product category in 1997?*"
* "*Which vendor invoices were entered into the system after midnight?*"

SQL provides a standardized, precise method for asking these questions and returning exact answers in table form.

## Why Learn SQL? The Accountant's Advantage

Data analytics is a process for using data to gain insights for business decision-making. While you can perform data analytics in spreadsheet software like Microsoft Excel, programming languages like SQL enforce a level of mental discipline that spreadsheets simply cannot match.

Mastering SQL gives accountants four crucial operational advantages:

1. **Explicit Logic:** In Excel, analytical steps are hidden inside individual cell formulas, buried across multiple worksheets, or executed through unrecorded mouse clicks. *SQL forces you to write a sequence of instructions.* This prevents haphazard trial-and-error and ensures you understand every step of your analytical process.
2. **Auditability & Reproducibility:** In professional accounting, an analysis that cannot be audited or reproduced by a peer, supervisor, or regulator is effectively useless. A SQL file serves as a permanent, readable audit trail. Months after an analysis is completed, another accountant can open your SQL file, review your exact logic, and re-execute the code with complete confidence.
3. **Data Integrity:** In a spreadsheet, data and calculations are combined; it is difficult to distinguish between raw and calculated data. This makes it easy to accidentally overwrite raw numbers or break formula references. SQL separates calculation logic (your code) from raw data storage (the database). Your query reads the database without modifying a single underlying record.
4. **Efficiency:** Accounting is defined by recurring reporting cycles -- monthly closes, quarterly reviews, and annual audits. In Excel, updating a report for a new period often requires tedious copying, pasting, and formula adjustments. With SQL, when new transaction data enters the database, you simply rerun your saved script to generate an updated report instantly.

Beyond these technical benefits, having genuine SQL proficiency on your resume provides significant "street cred" during job interviews. Accounting firm partners, corporate controllers, and recruiters consistently report that while basic Excel skills are expected of all accounting graduates, true database and query skills set job candidates apart as forward-thinking professionals who can work directly with enterprise data.

## SQL as a Universal Framework

Regardless of whether you are querying a SQL database, building an Excel Pivot Table, designing a dashboard in Tableau, or modeling data in Power BI, almost every data-oriented task in accounting boils down to **seven fundamental operations.**

Think of these seven operations as the universal blueprint for analytical thinking:

1. **Specifying the data source (FROM):** Identifying the primary database table containing the raw records you need.
2. **Merging (JOIN):** Linking related tables together (e.g., merging an *Orders* table to a *Customers* table so your report shows company names instead of customer IDs).
3. **Filtering (WHERE):** Restricting your analysis to specific rows of interest (e.g., filtering for transactions in a specific fiscal year or identifying accounts receivable that are more than 90 days past due).
4. **Selecting (SELECT):** Choosing the specific columns required for your final report, eliminating unnecessary data fields.
5. **Transforming (Operators & Functions):** Creating new calculated columns from existing data (e.g., multiplying unit price by quantity to compute line-item revenue, or calculating days elapsed between order and shipment).
6. **Aggregating (GROUP BY):** Summarizing detailed transaction-level data into high-level categories (e.g., computing total sales revenue per region or average transaction amount per employee).
7. **Sorting (ORDER BY):** Arranging the final results in a logical order (e.g., sorting highest expenses at the top or ordering transactions chronologically).

### Deconstructing a Real Accounting Query

To see how these seven operations come together, consider an example from the *Northwind Traders* sample database. Imagine your sales manager asks: *"List all customers located in Portland, Oregon, and show the total number of orders placed by each customer, sorted from highest to lowest."*

The SQL query below answers this exact business question.

```sql
SELECT
    C.CustomerID,
    C.CompanyName,
    COUNT(*) AS NumberOfOrders
FROM Orders AS O
INNER JOIN Customers AS C
    ON O.CustomerID = C.CustomerID
WHERE C.City = 'Portland'
GROUP BY C.CustomerID, C.CompanyName
ORDER BY NumberOfOrders DESC
```

## Summary and Looking Ahead

Learning SQL is not about memorizing commands. It is about training your mind to think sequentially about data.

Once you internalize these seven fundamental operations, your entire approach to data analysis will change. When you open Excel, Power BI, or Tableau, you will no longer click around haphazardly trying to guess a solution. Instead, you will pause, formulate your logical plan, identify the necessary transformations, and implement them cleanly.
